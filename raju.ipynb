```python
df.columns=['SNo','Category','State','Y2016','Y2017','Y2018','Share2018','Population','Rate2018']
```

```python
df.head()
```

```python
print(df['Category'].value_counts())
```

```python
d=df[~df['State'].str.contains('Total', case=False)]
print(d.shape)
d.head()
```

```python
print(d[['Y2016','Y2017','Y2018']].sum())
```

```python
y=d[['Y2016','Y2017','Y2018']].sum()
print(y)
plt.figure(figsize=(7,4))
sns.barplot(x=y.index, y=y.values, palette='Blues')
plt.title('Total Cases by Year')
plt.ylabel('Cases')
plt.show()
```

```python
plt.figure(figsize=(7,4))
plt.plot(['2016','2017','2018'], y.values, marker='o', color='red')
plt.title('Cases Over Years')
plt.ylabel('Cases')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Category', data=d, palette='Set2')
plt.title('State and UT Count')
plt.show()
```

```python
k=d.groupby('Category')[['Y2016','Y2017','Y2018']].sum()
print(k)
k.plot(kind='bar', figsize=(8,5))
plt.title('Cases State vs UT')
plt.tight_layout()
plt.show()
```

```python
t=d.sort_values('Y2018', ascending=False).head(10)
print(t[['State','Y2018']])
plt.figure(figsize=(10,5))
sns.barplot(data=t, x='State', y='Y2018', color='teal')
plt.xticks(rotation=45)
plt.title('Top 10 States 2018')
plt.tight_layout()
plt.show()
```

```python
t2=d.sort_values('Y2018').head(10)
plt.figure(figsize=(10,5))
sns.barplot(data=t2, x='State', y='Y2018', color='grey')
plt.xticks(rotation=45)
plt.title('Lowest 10 States 2018')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data=d.sort_values('Share2018', ascending=False).head(12), x='Share2018', y='State', color='purple')
plt.title('Share 2018')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data=d.sort_values('Rate2018', ascending=False).head(12), x='Rate2018', y='State', color='orange')
plt.title('Crime Rate 2018')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=d, x='Population', y='Y2018', hue='Category', size='Rate2018')
plt.title('Population vs Cases 2018')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=d, x='Y2017', y='Y2018', hue='Category')
plt.title('2017 vs 2018 Cases')
plt.show()
```

```python
h=d.set_index('State')[['Y2016','Y2017','Y2018']].head(15)
plt.figure(figsize=(8,8))
sns.heatmap(h, annot=True, fmt='.0f', cmap='YlOrRd')
plt.title('Cases Heatmap')
plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.heatmap(d[['Y2016','Y2017','Y2018','Share2018','Population','Rate2018']].corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Correlation')
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(data=d[['Y2016','Y2017','Y2018']])
plt.title('Year Case Spread')
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.histplot(d['Y2018'], bins=12, kde=True, color='green')
plt.title('2018 Cases Distribution')
plt.show()
```

```python
m=d.melt(id_vars=['State','Category'], value_vars=['Y2016','Y2017','Y2018'], var_name='Year', value_name='Cases')
plt.figure(figsize=(8,5))
sns.boxplot(data=m, x='Year', y='Cases', hue='Category')
plt.title('Cases by Year and Category')
plt.show()
```
