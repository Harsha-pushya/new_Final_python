```python
plt.figure(figsize=(11,6))
sns.barplot(data=df.sort_values('2021', ascending=False), 
            x='Crime Type', y='2021', color='steelblue')
plt.xticks(rotation=75)
plt.title('Cases 2021 by Crime Type')
plt.tight_layout()
plt.show()
```

```python
a=df.groupby('Category')['2021'].sum().sort_values(ascending=False)
print(a)
plt.figure(figsize=(8,5))
sns.barplot(x=a.index, y=a.values, palette='Set2')
plt.xticks(rotation=20)
plt.title('2021 Cases by Category')
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.pie(a, labels=a.index, autopct='%1.1f%%')
plt.title('Category Share 2021')
plt.show()
```

```python
y=df[['2016','2017','2018','2019','2020','2021']].sum()
print(y)
plt.figure(figsize=(8,4))
plt.plot(y.index, y.values, marker='o', color='darkred')
plt.title('Total Cases 2016 to 2021')
plt.ylabel('Cases')
plt.show()
```

```python
m=df.melt(id_vars=['Crime Type','Category'], value_vars=['2016','2017','2018','2019','2020','2021'], var_name='Year', value_name='Cases')
plt.figure(figsize=(9,5))
sns.lineplot(data=m, x='Year', y='Cases', hue='Category', marker='o', estimator=np.sum)
plt.title('Cases by Category Over Years')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Risk Level', data=df, palette='Reds', order=['Low','Medium','High'])
plt.title('Risk Level Count')
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.barplot(data=df, x='Category', y='Avg Loss Lakhs', hue='Risk Level')
plt.xticks(rotation=20)
plt.title('Average Loss by Category')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='2021', y='Avg Loss Lakhs', hue='Category', size='Percentage Share')
plt.title('Cases vs Loss')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x='Risk Level', y='Avg Loss Lakhs', data=df, palette='pastel', order=['Low','Medium','High'])
plt.title('Loss by Risk')
plt.show()
```

```python
h=df.set_index('Crime Type')[['2016','2017','2018','2019','2020','2021']]
plt.figure(figsize=(8,10))
sns.heatmap(h, cmap='YlGnBu')
plt.title('Crime Type by Year')
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.heatmap(df[['2016','2017','2018','2019','2020','2021','Percentage Share','Avg Loss Lakhs']].corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Correlation')
plt.show()
```

```python
t=df.sort_values('Percentage Share', ascending=False).head(8)
print(t[['Crime Type','Percentage Share','Risk Level']])
plt.figure(figsize=(9,5))
sns.barplot(data=t, x='Crime Type', y='Percentage Share', hue='Risk Level')
plt.xticks(rotation=45)
plt.title('Top Crime Share')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.histplot(df['2021'], bins=8, kde=True, color='teal')
plt.title('2021 Cases Distribution')
plt.show()
```
